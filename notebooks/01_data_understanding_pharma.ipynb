## **3. Findings:**

**1. Data loading:**
The original CSV files use ; as a separator. This was specified when loading the files so pandas could read them correctly.

**2. Table structure and keys:**

- `Laboratory`: 1005 rows × 55 columns, one row per batch.
- `Process`: 1005 rows × 35 columns, one row per batch.
- `Normalization`: 25 rows × 3 columns, one row per product code.
- `batch` is unique, and `Laboratory` and `Process` contain the same batches in the same order.
- All 25 product codes in `Laboratory` exist in `Normalization`.
- None of the three tables has duplicate rows.


**3. Data types (Laboratory):** pandas classifies columns by storage format, while the article classifies them by meaning, so the counts differ:

- **pandas:** 34 float64, 13 int64 and 8 object columns.
- **By meaning:** 45 numeric, 9 categorical and 1 identifier (`batch`). The article reports 44 numeric parameters; The file contains 2 numeric columns not described in the article’s Table 4: `tbl_max_diameter` and `fct_max_diameter`. This explains the difference between the 53 columns in the article and the 55 in the file.

*Issues found:*

- 7 categorical variables are stored as integers (`code`, `size, api_code`, `api_batch`, `smcc_batch`, `lactose_batch`, `starch_batch`). They are identifiers, so they will be treated as categorical and excluded from numeric calculations.
- 6 numeric API columns are stored as text (`api_water`, `api_total_impurities`, `api_l_impurity`, `api_ps01, api_ps05, api_ps09`) because their missing values are recorded as blank spaces instead of empty cells (see finding 4).
- `strength` contains 4 doses (5, 10, 20 and 40 mg) with an inconsistent format (5MG vs 20M). Only the numeric part will be kept.
- `start` contains month and year only (e.g. nov.18), with Slovenian month abbreviations (maj, avg, okt).

**4. Missing values:** Hidden missing values: the API columns stored as text contain blank-space strings that isnull() does not detect. After converting them to numeric, the real missing values are:
- **api_water:** 	26 (2.6 %) missing values.
- **api_total_impurities:**	67 (6.7 %) missing values.
- **api_l_impurity:**	370 (36.8 %) missing values.
- **api_ps01 / api_ps05 / api_ps09:**	26 (2.6 %) missing values each.

The missing values checked (`api_water`, `api_l_impurity`) all belong to API source 3 (481 batches, the main supplier). This matches the article. `api_l_impurity` is missing in 77% of this source’s batches, so its use in the analysis must be decided carefully.

Other missing values in **Laboratory:** `api_content` (2), `tbl_min_weight` and `tbl_max_weight` (10 each).
The real total of missing values in Laboratory is 563, not the 36 detected by isnull().

**Process:** the 108 missing values (18 batches × 6 columns) are all in the duplicated quality columns (see finding 6).

**Normalization:** no missing values.

The treatment of missing values will be defined in notebook 03.

**5. Time coverage (start):**
The data covers November 2018 to April 2021. No batches were produced in July 2020.

The number of batches per month varies widely (from 4 in April 2020 to 79 in June 2019), so time-based analysis will be done per batch (I-MR charts) instead of monthly averages. It still has to be verified that batch follows chronological order.

**6. Duplicated quality columns in Process:**
The 2 quality columns compared  in `Process` (`Drug release average (%)`, `Total impurities`) are identical to those in `Laboratory` (difference = 0 in the 987 comparable batches), the other 4 are assumed to be copies based on their names and values. They will be dropped, and the complete columns from `Laboratory` will be used.

**7. Descriptive statistics (Laboratory):**
- Dissolution values match the article (`dissolution_av` mean 90.6%, `dissolution_min` mean 85.6%), which confirms the data was loaded correctly.
- Hardness: `tbl_max_hardness` (batch 124) and `fct_min_hardness` (batch 214) have values of 0, which are missing values. In addition, the average hardness falls outside the min–max range in 479 batches (tablets) and 337 batches (film-coated tablets). Hardness columns are inconsistent, so they will be excluded, and tensile strength (`tbl_tensile`, `fct_tensile`) will be used instead.
- `tbl_rsd_weight` has an extreme value (9.90 vs 75th percentile of 1.17).
- `tbl_yield` and `batch_yield` have values slightly above 100%.
- Impurity values are concentrated at 0.05, which is probably the quantification limit. Their distribution is not normal, so Spearman correlation will be used.
- Weight, thickness and diameter vary widely because of the 4 doses, so they should be compared within each dose.

**8. Descriptive statistics (Process):**
- `main_CompForce` mean ranges from 2.46 to 17.63 with no impossible values. The range depends on the dose.
- `tbl_speed_mean` is concentrated around two set points (~100 and ~120), with one extreme value (271.7).
- `total_waste` is already normalized with batch size (per the article); `startup_waste` is not, so it will be normalized using `Normalization` before comparing batches.
- stiffness and ejection: zeros in _mean and _max (9 and 1 batches) mean the sensor did not record, so they are missing values. Zeros in _min (80 and 65 batches) probably reflect press stops, so _min columns will not be used as process descriptors.
- `pre_CompForce_mean` is ~0 for most batches; precompression is probably not used for many products.
- Extreme values in `tbl_speed_change`, `tbl_speed_0_duration`, `fom_change` and `SREL_production_max`. These variables are not part of the business questions.
- weekend: 55 batches (5.5%) had a weekend pause.

**9. Normalization:**
size in `Laboratory` matches `Batch Size (tablets)` in 23 of 25 products. For codes 14 and 16 it differs by a factor of 10 (4,800,000 in `Laboratory` vs 480,000 in `Normalization`). The normalization factor (4.8) is consistent with 480,000, so this value will be used.

**10. Column names:**
Some columns have inconsistent names: spaces, capital letters, parentheses and a typo in resodual_solvent. They will be renamed to snake_case in notebook 03. Product code will be renamed to code to match the other tables.

## **4. Data dictionary**

### Laboratory (1005 rows, one row per batch)
*Source: Žagar & Mihelič (2022), Table 4.*

| Column | Type | Description |
|---|---|---|
| **Genealogy data** | | |
| `batch` | Identifier | Index column; identifies every product batch number. |
| `code` | Categorical | Groups batches into product sub-families defined by product code. |
| `strength` | Categorical | Strength of the product, i.e. mg of Active Pharmaceutical Ingredient (API) per tablet (mg/unit). |
| `size` | Categorical | Target number of tablets produced per batch (tablets). |
| `start` | Date | Starting time of production. |
| `api_code`, `api_batch` | Categorical | API material code and batch number. |
| `smcc_batch`, `lactose_batch`, `starch_batch` | Categorical | Silicified microcrystalline cellulose (SMCC), lactose and starch batch numbers. |
| **Incoming raw materials (lab analysis)** | | |
| `api_water` | Numeric | Water content in API, measured with the loss on drying method (%). |
| `api_total_impurities`, `api_l_impurity` | Numeric | API total impurities and L impurity content, measured with HPLC (%). |
| `api_content` | Numeric | Active ingredient content in raw material, excluding impurities, water, etc. (%). |
| `api_ps01`, `api_ps05`, `api_ps09` | Numeric | API particle diameter at 10% (ps01), 50% (ps05) and 90% (ps09) cumulative volume (µm). |
| `lactose_water` | Numeric | Lactose water content, measured with the loss on drying method (%). |
| `lactose_sieve0045`, `lactose_sieve015`, `lactose_sieve025` | Numeric | Lactose particle size: % of weighted residual on the 0.045, 0.15 and 0.25 mm sieves (%). |
| `smcc_water` | Numeric | SMCC water content (%). |
| `smcc_td`, `smcc_bd` | Numeric | SMCC tap density (td) and bulk density (bd) (g/ml). |
| `smcc_ps01`, `smcc_ps05`, `smcc_ps09` | Numeric | SMCC particle diameter at 10% (ps01), 50% (ps05) and 90% (ps09) cumulative volume (µm). |
| `starch_ph` | Numeric | Starch pH value. |
| `starch_water` | Numeric | Starch water content, measured with the loss on drying method (%). |
| **Intermediate product quality (lab analysis)** | | |
| `tbl_min_thickness`, `tbl_max_thickness` | Numeric | Tablet core min and max thickness, measured during compression (mm). |
| `fct_min_thickness`, `fct_max_thickness` | Numeric | Film-coated tablet min and max thickness, measured after coating (mm). |
| `tbl_min_weight`, `tbl_max_weight` | Numeric | Tablet core min and max weight, measured during compression (mg). |
| `tbl_rsd_weight` | Numeric | Tablet core weight relative standard deviation (RSD), measured during compression (%). |
| `fct_rsd_weight` | Numeric | Film-coated tablet weight RSD, measured after coating (%). |
| `tbl_min_hardness`, `tbl_max_hardness`, `tbl_av_hardness` | Numeric | Tablet core min, max and average hardness, measured during compression (N). |
| `fct_min_hardness`, `fct_max_hardness`, `fct_av_hardness` | Numeric | Film-coated tablet min, max and average hardness, measured after coating (N). |
| `tbl_tensile`, `fct_tensile` | Numeric | Tensile strength: normalized hardness parameter for tablet cores and film-coated tablets. |
| `tbl_yield`, `batch_yield` | Numeric | Yield based on target quantity for the compression process (tbl) and the whole batch (batch) (%). |
| `tbl_max_diameter`, `fct_max_diameter` | Numeric | Not described in the article; description inferred from the column name. |
| **Final product quality (lab analysis)** | | |
| `dissolution_av` | Numeric | Average % of API released from the final tablet in 30 minutes (%). |
| `dissolution_min` | Numeric (target) | Minimum % of API released from the final tablet in 30 minutes (%). |
| `resodual_solvent` | Numeric | Residual solvent content in final product, measured with gas chromatography (%). |
| `impurities_total` | Numeric (target) | Total impurities content in final product, measured with HPLC (%). |
| `impurity_o` | Numeric | Impurity O content in final product, measured with HPLC (%). |
| `impurity_l` | Numeric | Impurity L content in final product, measured with HPLC (%). |



### Process (1005 rows, one row per batch)
Source: Žagar & Mihelič (2022).

| Column | Type | Description |
|---|---|---|
| **Identifiers** | | |
| `batch` | Identifier | Batch number; links to `Laboratory`. |
| `code` | Categorical | Product code; defines the product sub-family. |
| **Tablet press speed and stops** | | |
| `tbl_speed_mean` | Numeric | Mean tablet press speed, excluding stops (tablets/hour). |
| `tbl_speed_change` | Numeric | Number of tablet press speed changes, normalized with batch size. Many changes may indicate difficult material handling. |
| `tbl_speed_0_duration` | Numeric | Duration of press stops (speed = 0), normalized with batch size; weekends excluded. |
| `weekend` | Boolean | Whether the batch run included a weekend (yes/no). |
| **Waste** | | |
| `total_waste` | Numeric | Total rejected tablets per batch, normalized with batch size. |
| `startup_waste` | Numeric | Total rejected tablets during the press start-up (tablets). |
| **Filling device** | | |
| `fom_mean` | Numeric | Mean filling device speed, excluding stops (rpm). |
| `fom_change` | Numeric | Number of filling device speed changes during start-up. Many changes may indicate difficult material handling. |
| **Compression force and uniformity** | | |
| `SREL_startup_mean` | Numeric | Mean relative standard deviation of main compression force (SREL) during start-up; indicates tablet uniformity (%). |
| `SREL_production_mean` | Numeric | Mean SREL during production (%). |
| `SREL_production_max` | Numeric | Max SREL during production (%). |
| `main_CompForce mean` | Numeric | Mean main compression force during production (kN). A more constant force indicates a more homogeneous blend. |
| `main_CompForce_sd` | Numeric | Standard deviation of main compression force during production (kN). |
| `main_CompForce_median` | Numeric | Median main compression force during production (kN). |
| `pre_CompForce_mean` | Numeric | Mean pre-compression force during production (kN). Not regularly used for this product. |
| **Tablet fill and height** | | |
| `tbl_fill_mean` | Numeric | Mean tablet fill depth during production; defines the volume of material compressed (mm). |
| `tbl_fill_sd` | Numeric | Standard deviation of tablet fill depth during production; high values may indicate poor material flow (mm). |
| `cyl_height_mean` | Numeric | Mean cylindrical height of the tablet during production; controls tablet height and hardness (mm). |
| **Equipment** | | |
| `stiffness_mean`, `stiffness_max`, `stiffness_min` | Numeric | Mean, max and min bottom punch stiffness during production (N). |
| `ejection_mean`, `ejection_max`, `ejection_min` | Numeric | Mean, max and min tablet ejection force during production; a rise may indicate tablet sticking (N). |
| **Start-up phase** | | |
| `Startup_tbl_fill_maxDifference` | Numeric | Difference between max and min tablet fill depth during start-up (mm). |
| `Startup_main_CompForce_mean` | Numeric | Mean main compression force during start-up (kN). |
| `Startup_tbl_fill_mean` | Numeric | Mean tablet fill depth during start-up (mm). |
| **Final product quality (duplicated, will be dropped)** | | |
| `Drug release average (%)`, `Drug release min (%)` | Numeric | Copies of `dissolution_av` and `dissolution_min` from `Laboratory`. |
| `Residual solvent`, `Total impurities`, `Impurity O`, `Impurity L` | Numeric | Copies of `resodual_solvent`, `impurities_total`, `impurity_o` and `impurity_l` from `Laboratory`. |

All attributes are summaries of the tablet press time series, calculated per batch.

### Normalization (25 rows, one row per product code)
Source: Žagar & Mihelič (2022).

| Column | Type | Description |
|---|---|---|
| `Product code` | Categorical | Product code; links to `code` in `Laboratory` and `Process`. |
| `Batch Size (tablets)` | Numeric | Target number of tablets per batch for each product code (tablets). Matches `size` in `Laboratory` except for codes 14 and 16 (see finding 9). |
| `Normalisation factor` | Numeric | Batch size divided by 100,000. Used to normalize batch-size-dependent variables (e.g. `startup_waste`). |

The article does not describe this table; descriptions are based on the exploration in this notebook.
